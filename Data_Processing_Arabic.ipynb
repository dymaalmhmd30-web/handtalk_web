```python
import os

# عدّلي المسار إذا لزم
dataset1_path = "/content/dataset1/RGB ArSL dataset"

dataset2_path = "/content/dataset2/ArASL_Database_54K_Final/ArASL_Database_54K_Final"

def analyze_dataset(path, name):
    print(f"\n===== {name} =====")

    if not os.path.exists(path):
        print("❌ المجلد غير موجود!")
        return

    # جلب أسماء الفئات
    classes = sorted([d for d in os.listdir(path) if os.path.isdir(os.path.join(path, d))])

    print(f"📌 عدد التصنيفات: {len(classes)}\n")

    total_images = 0

    for cls in classes:
        cls_path = os.path.join(path, cls)
        img_count = len([f for f in os.listdir(cls_path) if os.path.isfile(os.path.join(cls_path, f))])
        total_images += img_count
        print(f"- {cls}: {img_count} صورة")

    print(f"\n📷 العدد الكلي للصور: {total_images}")

# تشغيل الفحص
analyze_dataset(dataset1_path, "Dataset 1")
analyze_dataset(dataset2_path, "Dataset 2")
```

```python
import os
import random
import matplotlib.pyplot as plt
from PIL import Image

# مسارات المجلدات الحقيقية عندك
dataset1 = "/content/dataset1/RGB ArSL dataset"
dataset2 = "/content/dataset2/ArASL_Database_54K_Final/ArASL_Database_54K_Final"

# الفئات اللي نبغا نطابقها
pairs = {
    "Hah_group": ["Hah", "Heh", "ha", "haa"],
    "Yeh_group": ["Yeh", "ya", "yaa"],
    "TehMarbuta_group": ["Teh_Marbuta", "toot"],
    "Kaf_group": ["Kaf", "kaaf"],
    "Beh_group": ["Beh", "bb"],
    "Qaf_group": ["Qaf", "gaaf"],
}


def show_samples(class_name, source_path, title):
    """عرض 3 صور لفئة محددة"""
    cls_path = os.path.join(source_path, class_name)
    if not os.path.exists(cls_path):
        print(f"❌ الفئة {class_name} غير موجودة في {source_path}")
        return

    images = [f for f in os.listdir(cls_path) if f.lower().endswith(("jpg","png","jpeg"))]

    if len(images) == 0:
        print(f"⚠️ لا يوجد صور داخل {class_name}")
        return

    sample_imgs = random.sample(images, min(3, len(images)))

    plt.figure(figsize=(10, 3))
    for i, img_name in enumerate(sample_imgs):
        img = Image.open(os.path.join(cls_path, img_name))
        plt.subplot(1, 3, i + 1)
        plt.imshow(img)
        plt.axis('off')
        plt.title(class_name)

    plt.suptitle(title)
    plt.show()


# عرض كل مجموعة
for group_name, classes in pairs.items():
    print(f"\n===== {group_name} =====")

    for cls in classes:
        # نبحث الفئة في dataset1 ثم dataset2
        path = dataset1 if os.path.exists(os.path.join(dataset1, cls)) else dataset2
        source = "Dataset 1" if path == dataset1 else "Dataset 2"

        print(f"عرض فئة: {cls}  (من {source})")
        show_samples(cls, path, f"{cls} samples")
```

# تعديل حجم ابعاد الصور في كلا المجموعتين وحفظها في ملفات جديده

```python
from PIL import Image
import os

def analyze_image_sizes(path, name):
    print(f"\n===== {name} =====")

    if not os.path.exists(path):
        print("❌ المجلد غير موجود!")
        return

    classes = sorted([d for d in os.listdir(path) if os.path.isdir(os.path.join(path, d))])

    all_sizes = []

    for cls in classes:
        cls_path = os.path.join(path, cls)
        for f in os.listdir(cls_path):
            file_path = os.path.join(cls_path, f)
            if os.path.isfile(file_path):
                try:
                    with Image.open(file_path) as img:
                        all_sizes.append(img.size)  # img.size يعطي (width, height)
                except:
                    print(f"⚠️ لم يتمكن من فتح الصورة: {file_path}")

    if all_sizes:
        widths, heights = zip(*all_sizes)
        print(f"عدد الصور: {len(all_sizes)}")
        print(f"أبعاد أول صورة: {all_sizes[0]} (عرض × طول)")
        print(f"متوسط العرض: {sum(widths)/len(widths):.2f}")
        print(f"متوسط الطول: {sum(heights)/len(heights):.2f}")
    else:
        print("لا توجد صور صالحة في هذا المجلد.")

# فحص كل Dataset
analyze_image_sizes(dataset1_path, "Dataset 1")
analyze_image_sizes(dataset2_path, "Dataset 2")
```

```python
from PIL import Image
import os

def resize_dataset(original_path, output_path, size=(200, 200)):
    if not os.path.exists(original_path):
        print("❌ المجلد الأصلي غير موجود!")
        return

    os.makedirs(output_path, exist_ok=True)

    classes = sorted([d for d in os.listdir(original_path) if os.path.isdir(os.path.join(original_path, d))])

    for cls in classes:
        cls_input_path = os.path.join(original_path, cls)
        cls_output_path = os.path.join(output_path, cls)
        os.makedirs(cls_output_path, exist_ok=True)

        for f in os.listdir(cls_input_path):
            input_file = os.path.join(cls_input_path, f)
            output_file = os.path.join(cls_output_path, f)
            if os.path.isfile(input_file):
                try:
                    with Image.open(input_file) as img:
                        img_resized = img.resize(size)
                        img_resized.save(output_file)
                except Exception as e:
                    print(f"⚠️ خطأ في الصورة {input_file}: {e}")

    print(f"✅ تم تغيير حجم جميع الصور في {original_path} إلى {size} وحفظها في {output_path}")

# أمثلة للاستخدام
resize_dataset(dataset1_path, "/content/dataset1_resized")
resize_dataset(dataset2_path, "/content/dataset2_resized")
```

```python
from PIL import Image
import os

def count_failed_images(original_path, resized_path):
    failed = 0
    success = 0

    classes = sorted([d for d in os.listdir(original_path) if os.path.isdir(os.path.join(original_path, d))])

    for cls in classes:
        orig_cls = os.path.join(original_path, cls)
        resized_cls = os.path.join(resized_path, cls)

        for f in os.listdir(orig_cls):
            orig_img = os.path.join(orig_cls, f)
            resized_img = os.path.join(resized_cls, f)

            # إذا ما تم إنشاء الصورة بعد التغيير → يعني فشلت
            if not os.path.exists(resized_img):
                failed += 1
                continue

            # نتأكد أن الأبعاد فعلاً 200×200
            try:
                with Image.open(resized_img) as img:
                    if img.size == (200, 200):
                        success += 1
                    else:
                        failed += 1
            except:
                failed += 1

    print(f"📌 الصور التي تم تغيير حجمها بنجاح: {success}")
    print(f"⚠️ الصور التي فشل تغيير حجمها: {failed}")

# استدعاء الدوال لكل مجموعتين
print("===== Dataset 1 =====")
count_failed_images(dataset1_path, "/content/dataset1_resized")

print("\n===== Dataset 2 =====")
count_failed_images(dataset2_path, "/content/dataset2_resized")
```

```python
from PIL import Image, ImageFile
import os

# مسار الصورة التالفة
broken_image_path = "/content/dataset1/RGB ArSL dataset/Yeh/Yeh_61.jpg"

def try_fix_image(img_path):
    print("🔧 محاولة إصلاح الصورة...")

    if not os.path.exists(img_path):
        print("❌ الصورة غير موجودة!")
        return False

    try:
        # نحاول تحميل الصورة بالسماح بالصور المقطوعة
        ImageFile.LOAD_TRUNCATED_IMAGES = True

        with Image.open(img_path) as img:
            # نحاول إجبار PIL على قراءة البيانات كاملة
            img.load()

            # حفظ الصورة من جديد بصيغة سليمة
            fixed_path = img_path.replace(".jpg", "_fixed.jpg")
            img.save(fixed_path, format="JPEG", quality=95)

        print(f"✅ تم إصلاح الصورة بنجاح وحفظها هنا:\n{fixed_path}")
        return True

    except Exception as e:
        print(f"⚠️ فشل إصلاح الصورة: {e}")
        return False


# تشغيل المحاولة
try_fix_image(broken_image_path)
```

```python
from PIL import Image
import os

def verify_resized_images(path):
    correct = 0
    incorrect = 0
    total = 0

    classes = sorted([
        d for d in os.listdir(path)
        if os.path.isdir(os.path.join(path, d))
    ])

    for cls in classes:
        cls_path = os.path.join(path, cls)

        for f in os.listdir(cls_path):
            img_path = os.path.join(cls_path, f)

            if not os.path.isfile(img_path):
                continue

            total += 1

            try:
                with Image.open(img_path) as img:
                    if img.size == (200, 200):
                        correct += 1
                    else:
                        incorrect += 1
                        print(f"⚠️ صورة ليست 200×200: {img_path} | الأبعاد: {img.size}")
            except:
                incorrect += 1
                print(f"❌ لم يتمكن من فتح الصورة: {img_path}")

    print(f"\n📌 المجلد: {path}")
    print(f"📷 عدد الصور الإجمالي: {total}")
    print(f"✅ الصور الصحيحة (200×200): {correct}")
    print(f"⚠️ الصور غير الصحيحة: {incorrect}")


# فحص كلا المجلدين
verify_resized_images("/content/dataset1_resized")
verify_resized_images("/content/dataset2_resized")
```

```python
import os
from PIL import Image
import matplotlib.pyplot as plt

# مسار مجلد الصور الأصلي
input_folder = "/content/drive/MyDrive/MyDatasets/dataset2_resized"
# مجلد الحفظ داخل الكولاب
output_folder = "/content/dataset2_rgb"
os.makedirs(output_folder, exist_ok=True)

# المرور على جميع الفئات والمجلدات الفرعية
for class_name in os.listdir(input_folder):
    class_path = os.path.join(input_folder, class_name)
    if not os.path.isdir(class_path):
        continue  # تخطى الملفات غير المجلدات

    # إنشاء مجلد جديد للفئة داخل الكولاب
    output_class_path = os.path.join(output_folder, class_name)
    os.makedirs(output_class_path, exist_ok=True)

    # المرور على الصور داخل الفئة
    for img_name in os.listdir(class_path):
        img_path = os.path.join(class_path, img_name)
        if not os.path.isfile(img_path):
            continue  # تخطى المجلدات الفرعية

        # فتح الصورة وتحويلها إلى RGB
        img = Image.open(img_path)
        img_rgb = img.convert("RGB")
        img_rgb.save(os.path.join(output_class_path, img_name))

# عرض مثال على صورة من أول فئة
example_class_path = os.path.join(input_folder, os.listdir(input_folder)[0])
example_img_path = os.path.join(example_class_path, os.listdir(example_class_path)[0])
example_img = Image.open(example_img_path)
example_img_rgb = example_img.convert("RGB")

plt.figure(figsize=(8,4))
plt.subplot(1,2,1)
plt.title("Original")
plt.imshow(example_img, cmap='gray' if example_img.mode=='L' else None)
plt.axis('off')

plt.subplot(1,2,2)
plt.title("Converted to RGB")
plt.imshow(example_img_rgb)
plt.axis('off')
plt.show()

print(f"All images are saved in {output_folder}")
```

```python
import os
import shutil

# ====== المسارات ======
dataset1_path = "/content/drive/MyDrive/MyDatasets/dataset1_resized"
dataset2_path = "/content/drive/MyDrive/MyDatasets/dataset2_resized_rgb"
output_path = "/content/merged_dataset"

os.makedirs(output_path, exist_ok=True)


# ====== helper لنسخ الصور ======
def copy_images(src, dst):
    os.makedirs(dst, exist_ok=True)
    for img in os.listdir(src):
        src_path = os.path.join(src, img)
        dst_path = os.path.join(dst, img)
        if os.path.isfile(src_path):
            shutil.copy(src_path, dst_path)

# ====== 1. دمج Dataset 1 (مع إعادة التسمية) ======
for folder in os.listdir(dataset1_path):
    old_label = folder
    new_label = label_mapping.get(old_label, None)

    if new_label is None:
        # احتمال نادر جداً لأنك أعطيتِ كامل mapping
        new_label = old_label.lower()

    src_folder = os.path.join(dataset1_path, folder)
    dst_folder = os.path.join(output_path, new_label)

    copy_images(src_folder, dst_folder)

# ====== 2. دمج Dataset 2 كما هي ======
for folder in os.listdir(dataset2_path):
    src_folder = os.path.join(dataset2_path, folder)
    dst_folder = os.path.join(output_path, folder.lower())

    copy_images(src_folder, dst_folder)

print("🎉 تم دمج البيانات بنجاح! المجلد النهائي جاهز:", output_path)
```

```python
# 1) Mount Google Drive
from google.colab import drive
drive.mount('/content/drive')

# 2) عدّل هذا المسار إلى مسار المجلد المطلوب داخل Drive
# مثال: /content/drive/MyDrive/dataset
ROOT_DIR = "/content/drive/MyDrive/merged_dataset_ArSL_rgb_New"

import os

# امتدادات الصور الشائعة
IMG_EXTS = {".jpg", ".jpeg", ".png", ".bmp", ".gif", ".tif", ".tiff", ".webp"}

def is_image_file(filename: str) -> bool:
    return os.path.splitext(filename.lower())[1] in IMG_EXTS

# 3) احصائيات عامة + تفصيل لكل مصنف (مجلد فرعي)
total_images = 0
subfolders = set()
images_per_folder = {}  # folder_path -> count

for dirpath, dirnames, filenames in os.walk(ROOT_DIR):
    # سجّل المجلدات الفرعية (مصنفات)
    for d in dirnames:
        subfolders.add(os.path.join(dirpath, d))

    # عدّ الصور داخل هذا المجلد
    img_count_here = sum(is_image_file(f) for f in filenames)
    if img_count_here > 0:
        images_per_folder[dirpath] = img_count_here

    total_images += img_count_here

# عدد المصنفات (بدون مجلد الجذر)
num_classes = len(subfolders)

print("✅ Folder:", ROOT_DIR)
print("📁 عدد المصنفات (المجلدات الفرعية):", num_classes)
print("🖼️ عدد الصور (داخل كل المجلدات):", total_images)

print("\n--- تفصيل الصور لكل مصنف (مجلد) ---")
for folder, cnt in sorted(images_per_folder.items(), key=lambda x: x[0]):
    # اسم مختصر بدل المسار الطويل
    short = folder.replace(ROOT_DIR, "").lstrip(os.sep)
    short = short if short else "(ROOT)"
    print(f"{short}: {cnt}")
```
